## Note to instructor who decides to read this code:
I'm labelling fources (F1, F2, F_r) when the units are grams. I know that may be confusing.
For computational purposes, I eleced to cut to the chase and just find the masses that students would need.
If you do the calculation with pen and paper (or whiteboard and marker), you'll notice the factor of 'g' goes away in the end anyways.
Thank you for your consideration

```python
import numpy as np
import random

def angle_difference(a1, a2):
    return min(abs(a1 - a2), 360 - abs(a1 - a2))

class Vector2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y
        
    def __str__(self):
        return f"({self.x}, {self.y})"
    
    def __repr__(self):
        return f"Vector2D(x={self.x}, y={self.y})"
    
    def magnitude(self):
        return np.sqrt(self.x**2 + self.y**2)
    
    def radians(self):
        return np.arctan2(self.y, self.x)
    
    def degrees(self):
         angle = 180*np.arctan2(self.y, self.x)/np.pi
         return angle
    
    def __add__(self, other):
        return Vector2D(self.x + other.x, self.y + other.y)
```

```python
groups = int(input("How many groups are there today? "))
```

```python
for i in range(groups):
    print(f"Group {i+1}:")
    while True:
        m1 = float(5.0*random.randint(1, 30))

        m2 = float(5.0*random.randint(1, 30))

        angle1 = random.randint(0, 90)

        angle2 = random.randint(91, 270)
        
        F1x = m1 * np.cos(angle1*np.pi/180)
        F2x = m2 * np.cos(angle2*np.pi/180)


        F1y = m1 * np.sin(angle1*np.pi/180)
        F2y = m2 * np.sin(angle2*np.pi/180)

        F1 = Vector2D(F1x, F1y)
        F2 = Vector2D(F2x, F2y)

        F_r = F1 + F2

        angle3 = (F_r.degrees() + 180)%360

        angle_1_3 = angle_difference(angle1, angle3)
        angle_2_3 = angle_difference(angle2, angle3)
        if  angle_1_3 > 10 and  angle_2_3 > 10 :
            break


    print(f"\tMass 1: {m1:.1f} grams.")
    print(f"\tMass 2: {m2:.1f} grams.")
     
    print(f"\tAngle 1: {angle1} degrees.")
    print(f"\tAngle 2: {angle2} degrees.")

    print(f"\tResultant mass: {F_r.magnitude():.1f} grams.")
    print(f"\tCounterbalance angle: {angle3:.0f} degrees.") 
```
